```python
# create two histograms on the data for frequency/distribution of distance and aadt.
import matplotlib.pyplot as plt

# Assuming the data is in a DataFrame called df with columns 'distance' and 'aadt'
plt.figure(figsize=(12, 5))

# Histogram for distance
plt.subplot(1, 2, 1)
plt.hist(df['Distance (KM)'], bins=20, color='blue', edgecolor='black')
plt.title('Distribution of Distance')
plt.xlabel('Distance (KM)')
plt.ylabel('Frequency')

# Histogram for aadt
plt.subplot(1, 2, 2)
plt.hist(df['2024 AADT'], bins=20, color='green', edgecolor='black')
plt.title('Distribution of AADT')
plt.xlabel('2024 AADT')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```
